## Visualizing Multi-Dimensional Tensors

A 2D tensor is simply a matrix laid out in table form. The first dimension or axis is the row, and the second dimension or axis is the column.

A 3D tensor, lets say $3 \times 5 \times 2$ can be visualized in the following ways:

First is to visualize it as it is printed by NumPy, i.e., we start with a $5 \times 2$ matrix, and then have 3 such matrices stacked on top of each other.

Second is to visualize it as a $3 \times 5$ matrix with each element of the matrix having $2$ elements.

The indexing will work in both ways.

```python
a = np.arange(3 * 5 * 2).reshape((3, 5, 2))
a
```

```python
a[1, 3, 0]
```

Now, lets visualize this the second way -

![visualization-1](../imgs/visualization-1.png)

Here, too, we see that selecting the `[1, 3]` element will get us the 2-value element `[16, 17]`, and selecting the `0`th element from this will give us `16`

Now, lets say that I have a 4D tensor $2 \times 3 \times 5 \times 2$. This can be visualized in the following ways -

First, the way it is printed, we have a block of $3$ stacks of $5 \times 2$ matrices, and $2$ such blocks.


Second, we have $2$ blocks of $3 \times 5$ matrices, each element of the matrix containing $2$ elements.

Third, we have a single $2 \times 3$ matrix, with each element being $5 \times 2$.

```python
a = np.arange(2 * 3 * 5 * 2).reshape((2, 3, 5, 2))
a
```

```python
a[0, 2, 4, 1]
```

Now, lets visualize it the second way with $2$ blocks of $3 \times 5$ matrices, each element of the matrix containing $2$ elements -

![visualization-2](../imgs/visualization-2.png)

To get to `[0, 2, 4, 1]`, I first get the `0`th block which gives me the $3 \times 5$ matrix. In there I choose the `[2, 4]`, which gives me the 2-valued element `[29, 29]`. Choosing the `[1]` element in there gives me the value $29$.

Finally, lets visualize it the third way, a single $2 \times 3$ matrix, with each element being $5 \times 2$.

![visualization-4](../imgs/visualization-4.png)

To get to `[0, 2, 4, 1]`, I first get the `[0, 2]` element in the $2 \times 3$ outer matrix. This gives me the inner $5 \times 2$ matrix. From here I choose the `[4, 1]` element.

Generally speaking, I can take visualize any $n$-dimensional tensor in $n - 1$ different ways. Lets take a final example of a 5D tensor of shape $3 \times 2 \times 3 \times 5 \times 2$. Here are the ways I can visualize this -

  * We start with $5 \times 2$ matrix, $3$ such matrices stacked on top of each other, $2$ such blocks of stacks of $3$, and $3$ super blocks each containing $2$ blocks.

  * We start with $3 \times 5$ matrix with each element having $2$ values. $2$ such matrices stacked on top of each other, and $3$ blocks of stacks of $2$.

  * We start with $2 \times 3$ matrix, each element of the matrix being a $5 \times 2$ matrix itself. We have $3$ such outer matrices stacked on top of each other.

  * Finally we have a single $3 \times 2$ matrix with each element being a $3 \times 5 \times 2$ tensor. Now this inner tensor can be visualized in 2 different ways.

Of course instead of arranging the numbers in 2D matrices, I can also arrange them in 3D cubes, e.g., one visualization can be cubes of $3 \times 5 \times 2$ arranged in a $3 \times 2$ grid.
