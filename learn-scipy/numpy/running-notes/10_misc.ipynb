### Histograms

The `counts, bins = np.histogram(a, bins=10, range=None, density=None, weights=None)` function will flatten the input provided, create the appropriate number of bins, and count the frequencies.

I can specify the number of bins, the range of the bins, whether to output a frequency count or a probability density number, the weight of each element of `a`. The function will output the counts (densities) and the bins. The number of bins will be one more than the counts because the bins will be $[b_0, b_1), [b_1, b_2), \cdots, [b_{n-1}, b_n]$.

There is a similar function in matplotlib that will calculate the histogram and plots it. This funciton also returns a tuple, but it is a 3-tuple of `(counts, bins, graphobj)`.

```python
mu, sigma = 2, 0.5
```

```python
v = rng.normal(mu, sigma, 10_000)
```

```python
(counts, bins) = np.histogram(v, bins=50)
```

```python
print(counts.shape)
print(bins.shape)
```

```python
for i in range(bins.shape[0] - 1):
    lower, upper = bins[i], bins[i + 1]
    count = counts[i]
    print(f"{lower:.3f} - {upper:.3f}: {count}")
```

```python
mid_points = 0.5 * (bins[:-1] + bins[1:])
mid_points
```

```python
plt.plot(mid_points, counts)
```

```python
counts, bins, _ = plt.hist(v, bins=50)
```

```python
for i in range(bins.shape[0] - 1):
    lower, upper = bins[i], bins[i + 1]
    count = counts[i]
    print(f"{lower:.3f} - {upper:.3f}: {count}")
```
