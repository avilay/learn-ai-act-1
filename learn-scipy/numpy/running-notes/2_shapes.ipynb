## Changing Shape

Three main functions for this -
  * `ndarray.reshape()` gives a new shape to an array without changing its data. 
  * `ndarray.resize()` changes the shape of the array in-place.
  * `ndarray.ravel()` will return the same data flattened in row-major form.

```python
a = np.arange(10).reshape(5, 2)
```

```python
b = a.reshape(2, 5)
```

```python
print(a)
print("---")
print(b)
```

```python
b[1, 2] = -1
```

```python
print(a)
print("---")
print(b)
```
