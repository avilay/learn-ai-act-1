## Stacking

For 2D arrays, it is conceptually easier to use `vstack()` and `hstack()`. `vstack()` will stack arrays on top of each other, or in other words, will stack along axis-0, and `hstack()` will stack arrays alongside each other, or in other words, will stack along axis-1.

`column_stack()` is best used for 1D arrays, where I have a bunch of column vectors and I want to stack them together to make a matrix. It is equivalent to `hstack()` in the 2D case.

For multi-dimensional arrays with more than 2 dimensions, it is best to use `concatenate()`.

When stacking 2 arrays, except for the axis they are being stacked on, all other axis should match, e.g., if I am stacking on axis-0, then two arrays can have mismatches on axis-0, but all other axes should match. Taking a more concrete example, if I want to stack $3 \times 4$ array with $3 \times 2$ array, I can only do this on axis-1 because that is the one that is mismatched. If I try to stack it along axis-0, it will fail because of axis-1 mismatch. On the other hand, I can stack a $3 \times 4$ array with $2 \times 4$ array on axis-0.

```python
v = rng.integers(-10, 10, 3)
v
```

```python
u = rng.integers(-10, 10, 3)
u
```

```python
np.column_stack((v, u))
```

```python
A = rng.integers(-10, 10, (3, 4))
A
```

```python
B = rng.integers(-10, 10, (3, 2))
B
```

```python
C = rng.integers(-10, 10, (2, 4))
C
```

```python
AC = np.vstack((A, C))
AC
```

```python
np.concatenate((A, C), axis=0)
```

```python
AB = np.hstack((A, B))
AB
```

```python
np.concatenate((A, B), axis=1)
```

```python
# This won't work because of mismatched dims
try:
    np.vstack((A, B))
except ValueError as ve:
    pprint(ve)
```

```python
X = rng.integers(-20, 20, (2, 3, 4))
X
```

```python
Y = rng.integers(-20, 20, (2, 4, 4))
Y
```

```python
XY = np.concatenate((X, Y), axis=1)
XY
```

## Splitting

This is conceptually opposite of stacking. It has the same three functions `vsplit()`, `hsplit()`, and `split()`. In addition to giving the array that I want to split, I also need to specify where I want the split. If I give a single scalar, it is interpreted as the number of splits. If I give an array, it is interpreted as the axis index at which I want the split.

```python
AC
```

```python
# Split AC along rows, s.t. row[3] is the start of the next split
np.vsplit(AC, [3])
```

```python
AB
```

```python
# Split AB into 3 equal splits
np.hsplit(AB, 3)
```

```python
XY
```

```python
np.split(XY, [3], axis=1)
```
