## Basic Operations

Unary operations such as computing the sum of an array act on the flat array by default. However, if I specify the `axis` parameter, it will act on that specific dimension. 

```python
b = np.arange(12).reshape(3, 4)
b
```

```python
np.sum(b)
```

```python
np.max(b)
```

Lets say I have a 2D matrix and I specify `axis=0` in the unary op, then it will first slice up the matrix along the specified dimension, in this case it will slice up the matrix into its rows. Then it will take the corresponding elements from each slice and construct the array input for the op, and finally run the op.

![op-axis-1](../imgs/op-axis-1.png)

In case of `sum` we end up calling -
  * `sum([0, 4, 8]) = 12`
  * `sum([1, 5, 9]) = 15`
  * `sum([2, 6, 10]) = 18`
  * `sum([3, 7, 11]) = 21`  

In case of `max` we end up calling -
  * `max([0, 4, 8]) = 8`
  * `max([1, 5, 9]) = 9`
  * `max([2, 6, 10]) = 10`
  * `max([3, 7, 11]) = 11`

One way this is explained in a lot of tutorials and such is that we are summing up the columns. I don't like that explanation because `axis=0` is supposed to be the rows, summing up columns leads to cognitive dissonance. I like my way of thinking in terms of slices and corresponding elements a lot more robust as it translates well to higher dimensions as well.

```python
np.sum(b, axis=0)
```

```python
np.max(b, axis=0)
```

Now, if I specify `axis=1`, it will first slice up the matrix along the columns, and then feed the corresponding elements in each slice to the op just like before.

![op-axis-2](../imgs/op-axis-2.png)

```python
np.sum(b, axis=1)
```

```python
np.max(b, axis=1)
```

```python
# c = rng.integers(1, 20, (3, 2, 3))

# fmt: off
c = np.array(
    [
        [
            [1, 4, 17], 
            [14, 4, 11]
        ],
        [
            [12, 18, 1],
            [10, 4, 12],
        ],
        [
            [9, 16, 18], 
            [16, 1, 2]
        ],
    ]
)
c
```

```python
c.shape
```

Lets consider a 3D tensor. When I specify `axis=0`, it will slice the tensor into 3 matrices, and then feed the corresponding elements -- in this case 3 elements each -- into the op.

![op-axis-3](../imgs/op-axis-3.png)

```python
np.sum(c, axis=0)
```

```python
np.max(c, axis=0)
```

The input array's dims were $3 \times 2 \times 3$, but after operating on the $0^{th}$ axis, the dimensions of the output are $2 \times 3$. The $0^{th}$ dimension has disappeared.

It gets interesting when I specify `axis=1`, now it will go through each element of the `0`th dimension and slice the element there -- in our case a 2D matrix -- along the rows. This will result in 3 sets of slices, because we have 3 elements in our `0`th dimension. The input will constructed for each set separately and fed to the op.

![op-axis-4](../imgs/op-axis-4.png)

```python
np.sum(c, axis=1)
```

```python
np.max(c, axis=1)
```

After operating on the $1^{st}$ axis, it will disappear and the output will be $3 \times 3$. This can become problematic if down the line the output is expected to have the same shape as the input. I can pass in the `keepdims=True` flag which will keep the axis that was...axed...with a size of $1$. This works well with broadcasting.

```python
d = np.sum(c, axis=1, keepdims=True)
d
```

```python
d.shape
```

Setting `axis=2` is similar except that it will slice the tensor into 3 sets of columns.

```python
np.sum(c, axis=2)
```

```python
np.max(c, axis=2)
```

Here are some common operations that work like described above:

  * `all()` and `any()`
  * `max`, `argmax()` and `min()`, `argmin()`
  * `average()` can be used to calculate weighted averages, `mean()`, `median()`, `std()`, `var()`
  * `sum()`, `prod()`

> Don't confuse `max()` with `maximum()`, `maximum()` takes two arrays as input and returns the element-wise maximum. Same with `min()` and `minimum()`.
