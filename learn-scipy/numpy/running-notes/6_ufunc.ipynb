## Universal Functions

A number of familiar math functions like $sin$, $cos$, $exp$, etc. are provided OOB by NumpPy and they operate on each element of the array. These are called [`ufunc`](https://numpy.org/doc/stable/reference/ufuncs.html) or universal functions in NumPy lingo.

```python
np.exp([1, 2, 3])
```
