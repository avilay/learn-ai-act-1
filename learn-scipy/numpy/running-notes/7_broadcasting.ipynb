# Broadcasting

Ref: https://numpy.org/doc/stable/user/basics.broadcasting.html

When I apply a binary function to 2 ND-Arrays, numpy will attempt to apply the function element-by-element. This means that both the input tensors should have the exact same shape, i.e., same number of dimensions, and each dimension having the same number of elements. Broadcasting techniques come into play when the shapes are not exactly the same. There are two ways in which we can have a shape mismatch.

  1. The number of dims is not the same.
  2. The number of elements for a dim do not match.

Broadcasting will first compare the dims starting from right to left. Additional dims are "padded" on the left of a tensor if needed, with a size of 1. Think of this as "unsqueezing" the tensor. Then again, starting from right to left, if the sizes of dims match, all is well. If they don't, then the block to the right of the mismatched dim is replicated. Otherwise broadcasting will fail with a `ValueError: operands could not be broadcast together with shapes <left shape>, <right shape>`. At the end of this process both the tensors should have the same shape.

Ofcourse all of this is happening conceptually. The actual implementation is rarely this wasteful in terms of memory.

```python
ones = np.ones((2, 3, 3, 2, 2))
```

```python
a = np.arange(2 * 3 * 3 * 2 * 1).reshape(2, 3, 3, 2, 1)
a
```

```python
ones * a
```

```python
b = np.arange(2 * 3 * 3 * 1 * 2).reshape(2, 3, 3, 1, 2)
b
```

```python
ones * b
```

```python
c = np.arange(2 * 3 * 1 * 2 * 2).reshape(2, 3, 1, 2, 2)
ones * c
```

```python
d = np.arange(2 * 1 * 3 * 2 * 2).reshape(2, 1, 3, 2, 2)
ones * d
```

```python
# Here the bigger shape is 2 x 3 x 3 x 2 x 2
# But e is                     3 x 3 x 2 x 2
# So an extra dim with size 1 will be first added to the left
# then the block to its right, essentially the entire tensor
# will be replicated twice.
e = np.arange(3 * 3 * 2 * 2).reshape(3, 3, 2, 2)
ones * e
```

I can "unsqueeze" an extra dim with size 1 anywhere in the multidim array by using the `np.newaxis`.

```python
x = np.arange(10).reshape((5, 2))
print(x.shape)
x
```

```python
y = x[:, np.newaxis, :]
print(y.shape)
y
```
