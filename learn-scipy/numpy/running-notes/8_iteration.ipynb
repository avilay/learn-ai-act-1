## Iteration

The default iterator will iterate over the first axis. If I want to iterate over each element, I can use the `ndarray.flat` property to get a flat iterator.

```python
b = np.fromfunction(lambda i, j, k: 100 * i + 10 * j + k, (2, 3, 4))
b
```

```python
for x in b:
    print(x)
    print("---")
```

```python
for x in b.flat:
    print(x)
```
