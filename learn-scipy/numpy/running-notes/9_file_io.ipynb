```python
# fmt: off
dt = np.dtype([
    ("time", [("min", np.int64), ("sec", np.int64)]), 
    ("temp", float)
])
```

```python
x = np.zeros((1,), dtype=dt)
x
```

```python
x["time"]["min"] = 10
x["temp"] = 98.25
x
```

```python
import tempfile

fname = tempfile.mkstemp()[1]
x.tofile(fname)
```

```python
np.fromfile(fname, dtype=dt)
```
